```python
import pandas as pd
import numpy as np
```

```python
assets = pd.read_csv("RTH/2025-04-01_assets.csv", parse_dates=[0])
assets.rename(columns={'Unnamed: 0': 'timestamp'}, inplace=True)

call_ask = pd.read_csv("RTH/2025-04-01_call_ask_20250416.csv", parse_dates=[0])
call_ask.rename(columns={'Unnamed: 0': 'timestamp'}, inplace=True)
call_bid = pd.read_csv("RTH/2025-04-01_call_bid_20250416.csv", parse_dates=[0])
call_bid.rename(columns={'Unnamed: 0': 'timestamp'}, inplace=True)

call_ask_eth = pd.read_csv("ETH/2025-04-01_call_ask_20250416.csv", parse_dates=[0])
call_ask_eth.rename(columns={'Unnamed: 0': 'timestamp'}, inplace=True)
call_bid_eth = pd.read_csv("ETH/2025-04-01_call_bid_20250416.csv", parse_dates=[0])
call_bid_eth.rename(columns={'Unnamed: 0': 'timestamp'}, inplace=True)
expiry = pd.to_datetime("2025-04-16")

put_ask = pd.read_csv("RTH/2025-04-01_put_ask_20250416.csv", parse_dates=[0])
put_ask.rename(columns={'Unnamed: 0': 'timestamp'}, inplace=True)
put_bid = pd.read_csv("RTH/2025-04-01_put_bid_20250416.csv", parse_dates=[0])
put_bid.rename(columns={'Unnamed: 0': 'timestamp'}, inplace=True)

put_ask_eth = pd.read_csv("ETH/2025-04-01_put_ask_20250416.csv", parse_dates=[0])
put_ask_eth.rename(columns={'Unnamed: 0': 'timestamp'}, inplace=True)
put_bid_eth = pd.read_csv("ETH/2025-04-01_put_bid_20250416.csv", parse_dates=[0])
put_bid_eth.rename(columns={'Unnamed: 0': 'timestamp'}, inplace=True)
expiry = pd.to_datetime("2025-04-16")
```

```python
strike_cols = [c for c in call_ask.columns if c != "timestamp"]

call_ask_long = call_ask.melt(id_vars="timestamp", value_vars=strike_cols, var_name="strike", value_name="ask")
call_bid_long = call_bid.melt(id_vars="timestamp", value_vars=strike_cols, var_name="strike", value_name="bid")

options_call = (call_bid_long.merge(call_ask_long, on=["timestamp", "strike"], how="outer"))
options_call["option_type"] = "C"
options_call["expiry"] = expiry
options_call["strike"] = options_call["strike"].astype(float)

# Mid-price
options_call["mid"] = 0.5 * (options_call["bid"] + options_call["ask"])
```

```python
put_ask_long = put_ask.melt(id_vars="timestamp", value_vars=strike_cols, var_name="strike", value_name="ask")
put_bid_long = put_bid.melt(id_vars="timestamp", value_vars=strike_cols, var_name="strike", value_name="bid")

options_put = (put_bid_long.merge(put_ask_long, on=["timestamp", "strike"], how="outer"))
options_put["option_type"] = "C"
options_put["expiry"] = expiry
options_put["strike"] = options_put["strike"].astype(float)

# Mid-price
options_put["mid"] = 0.5 * (options_put["bid"] + options_put["ask"])
```

```python
options = pd.concat([options_call, options_put], ignore_index=True)
options.head()
```
